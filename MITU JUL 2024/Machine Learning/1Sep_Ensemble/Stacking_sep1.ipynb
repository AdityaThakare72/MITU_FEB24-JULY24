```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
x = df[['Age', 'EstimatedSalary']]
y = df['Purchased']
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                   random_state= 0)
```

```python
x_train.shape
```

```python
y_train.shape
```

#### Build the model

```python
# import the classes
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
dt = DecisionTreeClassifier(random_state= 0)
nb = GaussianNB()
rf = RandomForestClassifier(random_state = 0)
log = LogisticRegression()
knn = KNeighborsClassifier(n_neighbors= 5)
svc = SVC()
```

```python
dt.fit(x_train, y_train)
nb.fit(x_train, y_train)
rf.fit(x_train, y_train)
log.fit(x_train, y_train)
knn.fit(x_train, y_train)
svc.fit(x_train, y_train)
```

```python
y_pred_dt = dt.predict(x_test)
y_pred_nb = nb.predict(x_test)
y_pred_rf = rf.predict(x_test)
y_pred_log = log.predict(x_test)
y_pred_knn = knn.predict(x_test)
y_pred_svc = svc.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print('LogR:', accuracy_score(y_test, y_pred_log)*100)
print('DT:', accuracy_score(y_test, y_pred_dt)*100)
print('KNN:', accuracy_score(y_test, y_pred_knn)*100)
print('NB:', accuracy_score(y_test, y_pred_nb)*100)
print('SVM:', accuracy_score(y_test, y_pred_svc)*100)
print('rf:', accuracy_score(y_test, y_pred_rf)*100)
```

### Stacking

```python
from sklearn.ensemble import StackingClassifier
```

```python
stack = StackingClassifier(estimators= [('LogR:',log),('DT:',dt),
                                    ('KNN:', knn),('NB:',nb),
                                    ('SVM:',svc)
                                   ],
                      final_estimator= rf)
```

```python
stack.fit(x_train, y_train)
```

```python
y_pred  = stack.predict(x_test)

accuracy_score(y_test, y_pred)
```

### Voting

```python
from sklearn.ensemble import VotingClassifier
```

```python
vot = VotingClassifier(estimators= [('LogR:',log),('DT:',dt),
                                    ('KNN:', knn),('NB:',nb),
                                    ('SVM:',svc),('rf:',rf)
                                   ])
```

```python
vot.fit(x_train, y_train)
```

```python
y_pred  = vot.predict(x_test)

accuracy_score(y_test, y_pred)
```
