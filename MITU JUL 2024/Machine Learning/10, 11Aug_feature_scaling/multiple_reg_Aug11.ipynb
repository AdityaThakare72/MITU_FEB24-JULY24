```python
import pandas as pd
```

```python
# data=> mtcars.csv
```

```python
df = pd.read_csv('mtcars.csv')
df.head()
```

```python
df.shape
```

```python
list(df.columns)
```

```python
# separate input and output

x = df[['cyl','disp','hp']]

y = df['mpg']
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(x, y)
```

```python
reg.score(x, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# predict on new values

cyl = 2
disp = 160
hp = 110
```

```python
reg.predict([[cyl, disp, hp]])
```
