```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('BostonHousing.csv')
df.head()
```

```python
x = df.drop('medv', axis = 1)
y = df['medv']
```

```python
# standard scaler
from sklearn.preprocessing import StandardScaler
sca = StandardScaler()
x_scaled = sca.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_scaled, y,
                                                   random_state=0,
                                                   test_size= 0.20)
```

#### linear reg

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)

print('training acc=>', lr.score(x_train, y_train))
print('testing acc=>', lr.score(x_test, y_test))
```

#### polynomial reg

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree= 2)
x_poly = poly.fit_transform(x_scaled)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_poly, y,
                                                   random_state=0,
                                                   test_size= 0.20)
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)

print('training acc=>', lr.score(x_train, y_train))
print('testing acc=>', lr.score(x_test, y_test))
```

#### Ridge reg

```python
from sklearn.linear_model import Ridge
ridge = Ridge(alpha= 18)
ridge.fit(x_train, y_train)

print('training acc=>', ridge.score(x_train, y_train))
print('testing acc=>', ridge.score(x_test, y_test))
```

#### Lasso

```python
from sklearn.linear_model import Lasso
lasso = Lasso(alpha= 0.2)
lasso.fit(x_train, y_train)

print('training acc=>', lasso.score(x_train, y_train))
print('testing acc=>', lasso.score(x_test, y_test))
```
