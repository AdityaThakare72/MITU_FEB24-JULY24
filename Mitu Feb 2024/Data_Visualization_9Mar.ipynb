```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([12,23,35,43,56,64,78,85,98])
y = np.array([9,15,23,34,41,56,65,78,89])
z = np.array([98,87,78,61,56,44,36,28,12])
```

### Line Plot

```python
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,y, marker = '*', mfc = 'red', ls = '--', lw = 2, ms = 7)
# marker => D, >, <, ^, o
```

```python
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,z, marker = 'o', mfc = 'black', ls = '--', lw = 2, ms = 7,
        color = 'red')
```

```python
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,y, marker = 'o', mfc = 'red', ls = '--', lw = 2, ms = 7,
        label = '2023')
plt.plot(x,z, marker = 'o', mfc = 'black', ls = '--', lw = 2, ms = 7,
        color = 'red', label = '2024')
plt.grid()
plt.legend()
```

#### Subplot

```python
plt.subplot(2,2,1)
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,y, marker = '*', mfc = 'red', ls = '--', lw = 2, ms = 7)
plt.grid()

plt.subplot(2,2,2)
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,z, marker = 'o', mfc = 'black', ls = '--', lw = 2, ms = 7,
        color = 'red')
plt.grid()
```

### Bar Graph

```python
courses = ['Python', 'Java', 'C', 'C++', 'R']
students = [60,56,50,53,40]
```

```python
plt.title('Student Enrollment')
plt.xlabel('Courses')
plt.ylabel('No of Students')
plt.bar(courses, students, color = 'black', width = 0.3,
       edgecolor = 'red', linewidth = 3, ls = '--', alpha = 1);
plt.grid()
```

### Scatter Plots

```python
plt.title('Scatter Plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.scatter(x, y, marker= '>')
plt.grid()
```

#### Pie chart

```python
plt.title('Pie Chart')
plt.pie(students, labels= courses, autopct = '%.2f%%', radius= 0.8,
       shadow= True, colors = ['red', 'green', 'blue', 'orange', 
                               'purple']);
```

```python
plt.figure(figsize=(10,10))

plt.subplot(2,2,1)
plt.title('Line plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.plot(x,y, marker = 'o', mfc = 'red', ls = '--', lw = 2, ms = 7,
        label = '2023')
plt.plot(x,z, marker = 'o', mfc = 'black', ls = '--', lw = 2, ms = 7,
        color = 'red', label = '2024')
plt.grid()
plt.legend()

plt.subplot(2,2,2)
plt.title('Student Enrollment')
plt.xlabel('Courses')
plt.ylabel('No of Students')
plt.bar(courses, students, color = 'black', width = 0.3,
       edgecolor = 'red', linewidth = 3, ls = '--', alpha = 1);
plt.grid()

plt.subplot(2,2,3)
plt.title('Scatter Plot')
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.scatter(x, y, marker= '>')
plt.grid()


plt.subplot(2,2,4)
plt.title('Pie Chart')
plt.pie(students, labels= courses, autopct = '%.2f%%', radius= 0.8,
       shadow= True, colors = ['red', 'green', 'blue', 'orange', 
                               'purple']);
```
