### Concate the dataframes

```python
df
```

```python
df.drop(['gender', 'cgpa'], axis = 1, inplace= True)
```

```python
df
```

```python
# create a dataframe of columns gender, cgpa, roll
# this roll column will be the same as of roll column of df
gender = ['m', 'm', 'm', 'm', 'f', 'f', 'm', 'f', 'm', 'm']
cgpa = df['marks'] / 10 + 1.05

newdf = pd.DataFrame({
    'roll':range(101,111),
    'gender': gender,
    'cgpa': cgpa
})
```

```python
newdf
```

```python
df
```

```python
df.columns
```

```python
newdf.columns
```

```python
# merge

pd.merge(df, newdf, on = 'roll')
```
