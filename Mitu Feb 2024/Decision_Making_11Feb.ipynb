```python
# take purchase and selling cost from the user
# find out if made profit or loss and how much?

purchase_cost = int(input('Purchase cost '))
selling_cost = int(input('Selling cost '))

if selling_cost > purchase_cost:
    print('Profit:', selling_cost - purchase_cost)
else:
    print('Loss:', purchase_cost - selling_cost)
```

```python
# take three numbers
# find out the largest number
# Not allowed to use 'and' or 'or'
# use nested if
num1 = int(input('Enter the 1st Number '))
num2 = int(input('Enter the 2nd Number '))
num3 = int(input('Enter the 3rd Number '))

if num1 > num2:
    if num1 > num3:
        print('Largest is', num1)
    else:
        print('Largest is', num3)
else:
    if num2 > num3:
         print('Largest is', num2)
    else:
        print('Largest is', num3)
```

```python
# 2016 = Leap
# 1997 = Not Leap
# 2000 = Leap
# 1900 = Not Leap
# take user input and check if its leap year or not

year = int(input('Enter the year '))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap century Year")
        else:
            print('Not Leap century year')
    else:
        print('Leap Year')
else:
    print('Not a Leap Year')
```

```python
'''
Read a number from user and do following
1. find its square
2. find cube
3. check positive or negative
'''
num1 = int(input('Enter the Number '))

print('Menu\n1.Square\n2.Cube\n3.positive or negative\n4.Even or odd')

ch = int(input('Enter the choice'))

if ch == 1:
    print(num1 * num1)
elif ch == 2:
    print(num1 ** 3)
elif ch == 3:
    if num1 > 0:
        print('positive')
    else:
        print('Negative')
        
elif ch == 4:
    if num1 % 2 == 0:
        print('Even')
    else:
        print('Odd')
else:
    print('Invalid Choice')
    
```
