```python
top = tk.Tk()

top.title('Sauare of a Number')
top.geometry('500x300')

data = tk.IntVar()     # integer variable

def show():
    num = data.get()    # getting the data
    sq = num * num
    result = tk.Label(top, text = 'Square is %i' %sq).place(x = 50, y = 200)

# label
l = tk.Label(top, text = 'Input').place(x = 20, y = 100)

# entry
e = tk.Entry(top, textvariable = data).place(x = 100, y = 100)

# button
b = tk.Button(top, text = 'Submit', command = show).place(x = 50,y=150)
    

top.mainloop()
```

```python
# read two numbers and find addition

# Read 2 numbers and find addition

top = tk.Tk()

top.title('Addition')
top.geometry('500x500')

data1 = tk.IntVar()
data2 = tk.IntVar()

def show():
    num1 = data1.get()
    num2 = data2.get()
    add = num1 + num2
    sub = num1 - num2
    mult = num1 * num2
    #tk.Label(top, text = 'Addition: %i' %add).place(x=50, y=200)
    from tkinter import messagebox
    messagebox.showinfo("Result","Addition: %i\n Sub: %i\n Mult: %i" %(add, sub, mult))
    
tk.Label(top, text = 'Enter: ').place(x=20, y=100)
tk.Entry(top, textvariable = data1).place(x=100, y=100)
tk.Entry(top, textvariable = data2).place(x=200, y=100)
tk.Button(top, text = 'Add', command = show).place(x=50, y=150)

top.mainloop()
```
