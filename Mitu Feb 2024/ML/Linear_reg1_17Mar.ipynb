```python
# Simple linear regression
```

```python
# input data = Height

x = [[151], [174], [138], [186], [128], [136], [179], [163], [152],
     [131]]
```

```python
# output data = Weight

y = [63, 81, 56, 91, 47, 57, 76, 72, 62, 48]
```

```python
x
```

```python
y
```

```python
# import the class
from sklearn.linear_model import LinearRegression
```

```python
# create an object 
lr = LinearRegression()
```

```python
lr.fit(x, y)
```

```python
# value of slope
lr.coef_
```

```python
# value of intercept
lr.intercept_
```

```python
# prediction

lr.predict([[190]])
```

```python
lr.predict([[200]])
```
