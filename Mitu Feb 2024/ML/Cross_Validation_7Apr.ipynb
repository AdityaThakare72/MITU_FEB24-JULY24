```python
# data: Salary_Data.csv
```

```python
import pandas as pd
```

```python
df = pd.read_csv('Salary_Data.csv')
```

```python
df.head()
```

```python
x = df['YearsExperience'].values
x = x.reshape(30,1)
y = df['Salary']
```

### holdout

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, 
                                                   random_state=0,
                                                   test_size= 0.25)
```

### build linear reg model

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
print('training:', lr.score(x_train, y_train))
print('testing:', lr.score(x_test, y_test))
```

### K fold Method

```python
from sklearn.model_selection import KFold
```

```python
x.shape
```

```python
kf = KFold(n_splits= 5, random_state= 0, shuffle= True)
```

```python
for train, test in kf.split(x, y):
    print(train, '\n', test)
```

### Extract the train and test data

```python
for train, test in kf.split(x, y):
    print('Train:', x[train].flatten())
    print('Test:', x[test].flatten())
```

### Apply Kfold

```python
for train, test in kf.split(x,y):
    lr.fit(x[train], y[train])
    print('Traning accuracy:', lr.score(x[train], y[train]))
    print('testing accuracy:', lr.score(x[test], y[test]), '\n')
```
