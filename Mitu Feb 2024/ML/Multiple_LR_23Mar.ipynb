```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('mtcars.csv')
```

```python
df.head(3)
```

```python
df.shape
```

```python
list(df.columns)
```

### Separate input and output data

```python
# input
x = df[['disp', 'hp', 'wt']]

# output
y = df['mpg']
```

### Build the model

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x, y)
```

```python
lr.score(x, y)
```

```python
disp = 225
hp = 110
wt = 3.2

# what is the mpg?
```

```python
lr.predict([[disp, hp, wt]])
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.figure(figsize=(16, 9))

plt.subplot(2,2,1)
plt.title('hp vs mpg')
plt.scatter(x['hp'], y)


plt.subplot(2,2,2)
plt.title('wt vs mpg')
plt.scatter(x['wt'], y)

plt.subplot(2,2,3)
plt.title('disp vs mpg')
plt.scatter(x['disp'], y);
```
